```python
# variables and types
price = 12.5 #float
count = 4 #integer
name = "Flagstaff" #string
print(price * count)
print(type(price), type(count), type(name))
```

```python
# lists and indexing
scores = [70, 85, 90, 65]
print(scores[0]) #70
print(scores[-1]) #65
print(scores[1:3]) #start:stop, stop is not inclusive
```

```python
# loop
total = 0
for s in scores:
  total = total + s
print(total / len(scores))
```

```python
# function
def average(values):
  return sum(values) / len(values)
print(average(scores))
```

```python
#list index like scores[0] gets the item, in this case the first item indexed at 0
#slice like scores[1:3] gets a list of items, in this case start:stop, thus stop is not inclusive,
#so this will get the 1st index or second item and the 2nd index or 3rd item
```

```python
#max_minus_min_values
def max_minus_min_values(values):
  return max(values) - min(values)
print(max_minus_min_values(scores))
```

```python
# Dictionary
home = {"city": "Flagstaff", "price": 45000, "beds": 3}
print(home["city"]) #prints Flagstaff
print(home["beds"]) #prints 3
home["beds"] = 4 #assigning beds from 3 to 4
print(home) #print the home dictionary. beds now will have the value of 4
```

```python
# if/elif/else
def grade(score):
  if score >= 90:
    return "A"
  elif score >= 80:
    return "B"
  else:
    return "C or below"
print(grade(85), grade(95), grade(60)) #grade 85 will return B, grade 95 will return A, and grade 60 will return C or below
```

```python
# a loop with a condition
count = 0
for s in scores:
  if s > 75:
    count = count + 1
print(count) #this will print 2 since there are two grades greater than 75
```

```python
# a list of dictionaries
homes = [
    {"city": "Flagstaff", "price": 450000},
    {"city": "Phoenix", "price": 380000},
    {"city": "Flagstaff", "price": 520000}
]
for h in homes:
  print(h["city"], h["price"]) # this will print all 3 lists
```

```python
def count_above(values, cutoff):
  count = 0
  for v in values:
    if v > cutoff:
      count = count + 1
  return count
print(count_above(scores, 75))
```

```python
def avg_price(homes, city):
  total = 0
  count = 0
  for h in homes:
    if h["city"] == city:
      total = total + h["price"]
      count = count + 1
  return total / count
print(avg_price(homes, "Phoenix"))
```
